# 필수 2. 평균 예측 baseline과 회귀 평가지표

## 배경

교안에서는 회귀 문제를 `MSE`, `RMSE`, `MAE`, `R²`로 평가할 수 있다고 배웠습니다. 또한 R²를 설명하면서 **아무 특성도 사용하지 않고 y의 평균값만 예측하는 기준 모델(baseline)**을 소개했습니다.

이번 문제에서는 새로운 모델 알고리즘을 사용하지 않고, **train의 SalePrice 평균을 모든 valid 데이터의 예측값으로 사용하는 baseline**을 만들어 네 가지 회귀 평가지표를 확인합니다.

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# TODO 1. y_train 평균으로 baseline 값을 만드세요.
baseline_value = Y_train.mean()

# TODO 2. valid 데이터 전체에 같은 baseline 값을 예측하도록 만드세요.
Y_vaild_pred = np.full(len(Y_vaild), baseline_value)

# TODO 3. MSE, RMSE, MAE, R²를 계산하세요.
# mean_squared_error는 실제값과 예측값의 차이를 제곱하여 평균mse를 계산한다. 
mse = mean_squared_error(Y_vaild, Y_vaild_pred)
rmse = np.sqrt(mse)

# mean_absolute_error는 차이의 절대값을 평균한 MAE를 계산.
mae = mean_absolute_error(Y_vaild, Y_vaild_pred)

# R2 : 평가 데이터의 평균을 예측하는 기준과 비교하여 R2값을 계산한다.
r2 = r2_score(Y_vaild, Y_vaild_pred)

# TODO 4. 결과를 출력하세요.
print(baseline_value)
print(f"MSE : {mse:,.2f}")
print(f"RMSE: {rmse:,.2f}")
print(f"MAE : {mae:,.2f}")
print(f"R^2: {r2:.6f}")
```

### 필수 2 답변 작성

- **Q1.** MSE는 무엇을 의미하나요?
: 실제값과 예측값의 오차를 제곱한 뒤 평균한 값(사전적 정의), 큰 오차가 없으면 숫자가 커짐.(해석할 떄, 이 정도가 나왔기 떄문에 큰 오차가 보입니다. 라고 해석할 줄 알아야 한다.) 
평균 제곱 오차(MSE): 각 예측 오차를 제곱한 뒤 평균한 값 이다. 

- **Q2.** RMSE는 무엇을 의미하며 MSE와 어떤 차이가 있나요?
MSE 평균 제곱 오차에 제곱근을 적용한 값으로 오차의 크기 비교를 위해 사용한다. 

- **Q3.** MAE는 무엇을 의미하나요?
평균 절대 오차로 오차에 절댓값을 씌워 양의 오차, 음의 오차가 상쇄되어 정확도가 잘못 측정되지 않도록 한다.
실제값과 예측값 차이의 절댓값을 평균한 값, 평균적으로 얼마나 벗어났는 지 보여줌.

- **Q4.** R²가 0에 매우 가깝다면 이 baseline을 어떻게 해석할 수 있나요?
: 특성을 사용하지 않고 평균값만 예측하는 기준 모델 수준과 다르지 않음. 
04-1 R2 가 음수나 0에 가깝다는 의미
: R2는 높을수록 좋은데, 음수나 0에 가깝다는 이야기는 모델링은 진행가능하나 
-> 모델 자체가 해당 데이터에 대한 설명력이 0에 가깝거나 음수로 표현된다.
-> 즉, 데이터를 충분히 설명할 수 있는 모델은 아니다. 

- **Q5.** RMSE가 MAE보다 크게 나오는 이유를 교안의 설명을 이용해 작성하세요.
RMSE는 평균 제곱 오차에 절댓값을 적용한 결과로 오차를 제곱하여 오차를 심각하게 보기 위함이 목적이다.
